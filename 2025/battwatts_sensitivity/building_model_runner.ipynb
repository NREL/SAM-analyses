```python
import csv
import os
import pandas as pd

# From building IDs, assemble weather file and associated paths
#building_id = ["46496","112157","111549","126745","305","442412","404322","448494","67263","168160","1652","467842"]
#latitude = ["33.56","35.14","32.87","39.81","41.88","44.85","40.68","39.99","45.59","35.22","44.47","42.95"]
#longitude = ["-86.75","-111.67","-117.15","-105.14","-71.02","-93.03","-74.17","-82.88","-122.6","-101.71","-73.15","-87.9"]
building_id = ["404322","448494","67263","168160","1652","467842"]
latitude = ["40.68","39.99","45.59","35.22","44.47","42.95"]
longitude = ["-74.17","-82.88","-122.6","-101.71","-73.15","-87.9"]
years = [2018, 2019, 2020, 2021, 2022, 2023]

for i, file_lat in enumerate(latitude):
    for year in years:

        building_path = "building_models\\bldg" + building_id[i] + "-up00\\"
        building_filename = building_path + "bld" + building_id[i] +".idf"
        weather_name = building_path + "weather_data\\" + 'nsrdb_' + str(file_lat) + '_' + str(longitude[i]) + '_nsrdb-GOES-aggregated-v4-0-0_60_' + str(year) + ".epw"     

        prefix = building_path + str(year)

        # Run E+
        print(os.popen("energyplus -w " + weather_name + " -p  " + prefix + " " + building_filename).read())

        # Extract load profile in J, convert to kW

        df = pd.read_csv(prefix + 'mtr.csv')
        data = df["Electricity:Facility [J](Hourly)"]
        kwh = data * 2.7777777777778E-7

        filename = prefix + "-" + building_id[i] + "_kw_load.csv"

        load_file = open(filename, 'w', newline='\n', encoding='utf-8')
        writer = csv.writer(load_file, delimiter=',', lineterminator='\n')
        writer.writerow(['kW'])

        for l in kwh:
            writer.writerow([str(l)])

        load_file.close()
```
